### jdaviz + JupyterLite!

```python
import json
import jdaviz as jd
from astropy.io import fits
from sidecar import Sidecar
from mast_table import MastTable
from astropy.table import Table
```

```python
viz = jd.App()

image = fits.open('HorseHead.fits', memmap=False)
viz.load(image, loader='object', format='Image')

sidecar_jdaviz = Sidecar(title='jdaviz', anchor='split-right')

with sidecar_jdaviz:
    display(viz._app)

# observations = Table.read('cached_query.ecsv')

with open('cached_query.json', 'r') as cached_query:
    observations = Table(json.load(cached_query))

with Sidecar(title='mast-table', anchor='split-bottom', ref=sidecar_jdaviz):
    display(MastTable(observations))
```
